```python
import pandas as pd
```

```python
df=pd.read_csv("/content/github 1 .csv")
df.info()
```

```python
df.duplicated().sum()
```

```python
df.describe()
```

```python
df= df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df[df[['Attendance_Percent', 'Sleep_Hours_Per_Day', 'Assignment_Completion_Percent']].isnull().any(axis=1)]
```

```python
median_attendance = df['Attendance_Percent'].median()
df['Attendance_Percent'] = df['Attendance_Percent'].fillna(median_attendance)
df
```

```python
median_sleep = df['Sleep_Hours_Per_Day'].median()
df['Sleep_Hours_Per_Day'] = df['Sleep_Hours_Per_Day'].fillna(median_sleep)
df
```

```python
median_assign = df['Assignment_Completion_Percent'].median()
df['Assignment_Completion_Percent'] = df['Assignment_Completion_Percent'].fillna(median_assign)
df
```

```python
df.isnull().sum()
```

```python
high_sleep = df[df['Sleep_Hours_Per_Day'] > 7]['Final_Score'].mean()
low_sleep = df[df['Sleep_Hours_Per_Day'] <= 5]['Final_Score'].mean()

high_sleep, low_sleep
```

```python
high_study = df[df['Study_Hours_Per_Week'] >= 15]['Final_Score'].mean()
low_study = df[df['Study_Hours_Per_Week'] < 15]['Final_Score'].mean()

high_study, low_study
```

```python
high_att = df[df['Attendance_Percent'] >= 85]['Final_Score'].mean()
low_att = df[df['Attendance_Percent'] < 85]['Final_Score'].mean()

high_att, low_att
```

```python
import matplotlib.pyplot as plt
```

```python
categories = ['High Study (15+ hrs)', 'Low Study (<15 hrs)']

scores = [high_study, low_study]

plt.bar(categories, scores)
plt.title("How study hours affect scores of students?")
plt.xlabel("Study Tier")
plt.ylabel("Average Final Score")

plt.show()
```

```python
plt.scatter(df['Study_Hours_Per_Week'], df['Final_Score'], color='purple', alpha=0.6)


plt.title("Relationship Between Weekly Study Hours and Final Score")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Final Score")

plt.show()
```
